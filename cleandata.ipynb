```python
## check duplicate records
df.duplicated().sum()
```

```python
## drop duplicate 
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
## replace invalid values 
## replace common invalid values with  nan 
df.replace(['N/A','NULL',''],np.nan,inplace=True)
```

```python
## remove extra spaces 
df=df.apply(lambda x:x.str.strip() if x.dtype=='object' else x)
```

```python
## standerdize text format
## convert  text into  proper case
df['Customer_Name']=df['Customer_Name'].str.title()
df['City']=df['City'].str.title()
df['State']=df['State'].str.title()
```

```python
## validate email addresses
## keep only  valid email records

df=df[df['Email'].str.contains('@',na=False)]
```

```python
## convert data cloumns
df['Order_Date']=pd.to_datetime(df['Order_Date'],errors='coerce')
df['Delivery_Date']=pd.to_datetime(df['Delivery_Date'],errors='coerce')
```

```python
## convert numeric column 
cols=['Qty','Unit_Price','Discount']

for c in cols:
    df[c]=pd.to_numeric(df[c],errors='coerce')
```

```python
## remove invalid quantity
## remove records having nagtive or zero qunatity

df=df[df['Qty']>0]
```

```python
## fill missing values
df['Discount']=df['Discount'].fillna(0)

df['Phone']=df['Phone'].fillna('Unknown')

df['Delivery_Date']=df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
## feature engineering
## total sales

df['sales']=df['Qty']*df['Unit_Price']
```

```python
df
```

```python
## netamount
df['Net_amount']=df['sales']-(df['sales']*df['Discount']/100)
```

```python
df
```

```python
## profit (example: 20%) 
df['Profit']=df['Net_amount']*0.20
```

```python
df
```

```python
## order month  

df['Month']=df['Order_Date'].dt.month_name()
```

```python
## order year
df['year']=df['Order_Date'].dt.year
```

```python
df
```

```python
## weekday

df['Weekday']=df['Order_Date'].dt.day_name()
```

```python
df
```

```python
## final data validation
df.info()
```

```python
## export  clean data
df.to_csv('Clean_Ecommerce.csv',index=False)
```

```python
## import data into mysql

from sqlalchemy import create_engine

# XAMPP connection string (localhost, root user, no password by default)
# Apne database ka naam yahan 'your_database_name' ki jagah likhein
engine = create_engine("mysql+pymysql://root:@localhost/Ecommerce")

# DataFrame ko directly MySQL table mein export karein
df.to_sql(
    name="clean_ecommerce",  # Yeh MySQL mein table ban jayegi
    con=engine,
    if_exists="replace",  # Options: 'fail', 'replace', 'append'
    index=False,
)

print("Data successfully imported to XAMPP MySQL!")
```
